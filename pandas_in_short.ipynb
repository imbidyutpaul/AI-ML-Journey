# **Pandas**

Pandas is an open-source Python library used for data manipulation, data analysis, and data cleaning.

Pandas helps us to read, organize, filter, modify, and analyze datasets easily.


#  Why do we use Pandas?


 1.Data Reading: Read CSV, Excel, and JSON files.                                                     2.Data Cleaning: Handle missing values and duplicate values.                                         3.Data Manipulation: Add, delete, and modify columns.                                                4.Data Filtering: Filter specific rows and columns.                                                  5.Data Analysis: Calculate mean, median, mode, etc.                                                  6.Data Exploration: Understand the structure and distribution of data.

# **DataFrame**

**A DataFrame is a two-dimensional tabular data structure consisting of rows and columns.**

```python
# Create a dictionary
data = {
    "Name": ["Rahul", "Amit", "Priya"],
    "Age": [22, 24, 21],
    "Marks": [85, 90, 95]
}

# Create a DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
print(df)
```

# **Selecting Rows and Columns**

```python
# Display the first column
df.iloc[:, 0]

```

```python
# Display the first three columns
df.iloc[:, 0:3]

```

```python
# Display the first five rows
df.iloc[0:5]

```

```python
# Select columns by name
df[df.columns[0:3]]

```

```python
# Select rows using loc
df.loc[0:4]

```

```python
# Select rows and columns using loc
df.loc[0:4, df.columns[0:3]]

```

```python
# Select a single value using at
df.at[0, df.columns[0]]
```

# **Data Filtering**

```python
# Filter students whose age is greater than 20
df[df["age"] > 20]
```

```python
# Select only name and age columns
df.loc[df["age"] > 20, ["name", "age"]]

```

```python
df.loc[df["age"] > 20, ["gender", "age"]]
```

```python

df.loc[df["age"] > 20, ["age","total_lectures"]]
```

```python
# Filter students whose age is greater than 20 and gender is Male
df.query("age > 20 and gender == 'Male'")

```

```python
# Filter students whose gender is Male
df[df["gender"].isin(["Male"])]
```

```python
# Filter students whose gender is Female
df[df["gender"].isin(["Female"])]
```
